## 1. Saisie des deux series
Remplacez les listes par vos données BCEAO / INS-CI (datees, sources a citer).


```python
import numpy as np
import matplotlib.pyplot as plt

# A COMPLETER : vos données
X = np.array([])  # ex : croissance M2 (%)
Y = np.array([])  # ex : inflation (%)
```

## 2. Nuage de points


```python
plt.figure(figsize=(8, 5))
plt.scatter(X, Y, color='#1C2B5E')
plt.xlabel('X'); plt.ylabel('Y'); plt.title('Nuage de points X-Y')
plt.show()
```

## 3. Droite d'ajustement, R2, correlation


```python
if len(X) >= 2:
    a, b = np.polyfit(X, Y, 1)
    r = np.corrcoef(X, Y)[0, 1]
    r2 = r ** 2
    print(f'Equation : Y = {a:.2f} X + {b:.2f}')
    print(f'R2          : {r2:.2f}')
    print(f'Correlation : {r:.2f}')
else:
    print('Saisissez au moins 2 points dans X et Y.')
```

```python
if len(X) >= 2:
    plt.figure(figsize=(8, 5))
    plt.scatter(X, Y, color='#1C2B5E', label='Observations')
    xs = np.linspace(X.min(), X.max(), 50)
    plt.plot(xs, a * xs + b, color='#2645C8', label='Droite d\'ajustement')
    plt.legend(); plt.title('Regression lineaire'); plt.show()
```
